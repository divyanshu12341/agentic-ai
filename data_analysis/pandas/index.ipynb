```python
# Pandas 
# It is widely used in data-analysis and data-cleaning 
# Provides primary two data structures:Series and dataframe 
# Series is one dimensional array like object 
# DataFrame is two dimensional ,size-mutable and potentially heterogenous 
# tabular data structure with labeled axis(row and columns)
import pandas as pd 
data = [1,2,3,4,5]
series = pd.Series(data)
series
# 0 1 2 3 4 5 is basically indexes

```

```python
# Series from dictionary elements
data = {'a':1,'b':2,'c':3}
series_dict = pd.Series(data)
print(series_dict)

```

```python
data = [10,20,30]
index = ['a','b','c']
pd.Series(data,index)
```

```python
#DataFrame 
# Create dataframe from dictionary of list 
data = {
    'name':['Divyanshu','Tanmay','Vansh','Divy','Divyesh'],
    'age':[25,24,24,27,26],
    'city':['Bangalore','New York','Chicago','Florida','Las Vegas']
}
df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
# Create dataframe from list of dictionaries
data = [
    {'Name':'Divyanshu','age':26,'city':'Jaipur'},
    {'Name':'Divy','age':27,'city':'Dubai'},
    {'Name':'Darshan','age':28,'city':'Bangalore'}
]
df = pd.DataFrame(data)
df
```
