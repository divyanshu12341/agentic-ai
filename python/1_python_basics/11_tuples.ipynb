```python
# Tuples are ordered collection of items that are immutable 

# Creating a tuple 
empty_tuple = ()
print(empty_tuple)
```

```python
lst = list()
print(type(lst))
tpl = tuple()
print(type(tpl))
```

```python
numbers = tuple([1,2,3,4,5,6])
numbers
```

```python
mixed_tuple = (1,"Hello World",3.14,True)
print(mixed_tuple)
```

```python
# Accessing tuple elements
numbers
numbers[0]
```

```python
numbers[0:4]
numbers[::-1]
```

```python
# Tuple operations: 

#Concatenation which is combining two tuples 
concatenation_tuple = numbers + mixed_tuple
print(concatenation_tuple)
```

```python
mixed_tuple*3 #means tuple gets appended three times 
```

```python
# Immutable nature of tuples 
lst = [1,2,3,4,5]
print(lst)
lst[1] = 'Krissh'
print(lst)

numbers[1] = 'krish' #throws an error 
```

```python
#Tuple methods 
# no of times that particular el is present
print(numbers.count(1))
# index of that particular el 
print(numbers.index(3))
```

```python
## Packing and unpacking tuples 
packed_tuples = 1,"Hello",3.14
print(packed_tuples)

# unpacking tuples 
a,b,c = packed_tuples 
print(a,b,c)
```

```python
## Unpacking with * 
numbers = (1,2,3,4,5,6)
first,*middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
## Nested tuple 
## Nested list 
lst = [[1,2,3,4],[6,7,8,9],[1,"Hello",3.14]]
lst[0][3]

#nested tuple 
nestedtpl= ((1,2,3),("a","b","c"),(True,False))
print(nestedtpl[0][2])
print(nestedtpl[1][2])
print(nestedtpl[2][1])

#iterating over nested tuples 
for sub_tuple in nestedtpl:
    for items in sub_tuple:
        print(items)

```
