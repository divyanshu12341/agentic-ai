```python
#Lambda functions 
#Small anonymous functions defined using lambda keyword.
#Any no of arguments but only one expression

addition = lambda a,b:a+b
addition(5,10)
```

```python
even = lambda a:a%2
even(10)
```

```python
addition1 = lambda a,b,c:a+b+c
addition1(10,20,30)
```

```python
## map
numbers = [1,2,3,4,5,6]
sq_num=map(lambda num:num*num,numbers)
sq_num = list(sq_num)
print(sq_num)
```
