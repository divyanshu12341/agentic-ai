```python
## Sets store collection of unordered unique items
# Unordered means elements does not follow a specific order
#  
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
 
```

```python
my_empty_set = set()
print(my_empty_set)
```

```python
my_nonempty_set = set([1,2,3,4,5,6,5,4,6])
print(my_nonempty_set)
```

```python
# Basics set operation: 

#1) Adding and removing elements
my_nonempty_set.add(7)
my_nonempty_set
```

```python
#2) removing elements from set
my_nonempty_set.remove(3)
print(my_nonempty_set)
```

```python
my_nonempty_set.remove(10)
```

```python
my_nonempty_set.discard(10)
print(my_nonempty_set)
```

```python
# Pop method 
# Removes first element 
removed_element = my_nonempty_set.pop()
print(removed_element)
print(my_nonempty_set)
```

```python
#clear all the elements 
my_nonempty_set.clear()
print(my_nonempty_set)
```

```python
# Set membership test
my_nonempty_set = {1,2,3,4,5}
print(3 in my_nonempty_set)
print(10 in my_nonempty_set)
```

```python
# Mathematical operations 
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

#Union 
union_set = set1.union(set2)
print(union_set)

#Intersection 
intersection_set = set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)

#Difference 
# removes common elements
diff_set = set1.difference(set2)
print(diff_set)
```

```python
## Symmetric difference 
set1.symmetric_difference(set2)
```

```python
set1 = {1,2,3}
set2 = {3,4,5}
# is set1 a parent set of set2
set1.issubset(set2)
set1.issuperset(set2)
```
