```python
numbers = [1,2,3,4,5,6,7,8]
doubled_num = list(map(lambda num:num*2,numbers))
doubled_num
```

```python
# can we map multiple iterables 
numbers1 = [1,2,3]
numbers2 = [4,5,6]
added_num = list(map(lambda x,y:x+y,numbers1,numbers2))
```

```python
# map() to convert list of strings to integers
string_numbers = ['1','2','3','4','5']
int_numbers = list(map(lambda num:int(num),string_numbers))
int_numbers
```

```python
# map() to convert words to upper case
words = ['apple','banana','cherry']
upper_word = list(map(lambda str:str.upper(),words))
upper_word
```

```python
# getting specific property from list of objects
def get_name(person):
    return person['name']

people = [
    {'name':'Krish','age':32},
    {'name':'Jack','age':40}]
print(list(map(get_name,people)))

 
```
