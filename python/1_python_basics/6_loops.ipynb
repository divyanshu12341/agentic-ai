```python
## for loop 
## Last no is not included in the range 
for i in range(5):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
# range(start,stop,step-size)
for i in range(1,10,2):
    print(i)

for i in range(10,1,-1):
    print(i)
```

```python
# strings 
str = "Divyanshu jain"
for i in str:
    print(i)
```

```python
## while loop 

count = 0
while count<5:
    print(count)
    count = count+1
```

```python
count = 0
while count%2==0:
    print(count)
    count = count+1
```

```python
# loop control statements 
# break 
# break statement used to terminate the loop permaturely
for i in range(10):
    if i == 5:
        break
    print(i)
    
```

```python
#continue 
# skips current iteration and continues with the next 
for i in range(10):
    if i%2 ==0:
        continue
    print(i)
```

```python
# pass 
# it is a null operation and does nothing 
for i in range(5):
    if i==3:
        pass
    print(i)
```

```python
# Nested loops 
# Loop inside a loop 

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
# Examples: Calculate sum of first n natural no using while and for loop
n = 10
i = 0
sum = 0
while(i<n):
    sum = sum+i
    i = i+1
print(sum)
```

```python
sum = 0
for i in range(11):
    sum +=i
print(sum)
```

```python
# Prime no between 1 and 100
for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i ==0:
                break;
        else:
            print(num)
```
