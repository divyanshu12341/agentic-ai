```python
# Dictionaries are unordered collection of items 
# Here data is stored in key-value pairs 
# key always must be unique and immutable while values can be of any type

# Creating dictionaries 
empty_dictionary = {}
print(type(empty_dictionary))
```

```python
empty_dict = dict()
print(type(empty_dict))
```

```python
student = {
    "name":"Divyanshu",
    "age":26,
    "grade":26,    
}
student
type(student)
```

```python
# Single key is always used  
student = {
    "name":"Divyanshu",
    "age":32,
    "name":24
}
print(student)
```

```python
# Accessing dictionary elements 
student = {
    "name":"Divyanshu",
    "age":32,
    "Grade": "A"
}
student["Grade"]
student["age"]

# Accessing using get method 
print(student.get('Grade'))
# If key not present then we get None if we use get method 
print(student.get('last_name'))
print(student.get('last_name','Not available'))
```

```python
# Modifying dictionary elements
 # Dictionary are mutable 
 # You can add,update or delete elements 
print(student) 
student["age"] = 33   ## update value for key
print(student)
student["address"] = "India" ## added new key and value
print(student)

del student["address"] #delete key and value pair 
student
```

```python
## Dictionary methods 
keys = student.keys() ##get all the keys
print(keys)
values = student.values() #get all the values 
print(values)

items = student.items() #all key value pairs 
items
```

```python
## Shallow copy 
student_copy = student
student_copy    
student_copy['name'] = 'DJ'
student
```

```python
student_copy1 = student.copy() ## Shallow copy 
print(student_copy1)
```

```python
student["name"] = "Divyanshu"
student_copy1
```

```python
## Iterating over dictionaries 
## You can use loops to iterate over dictionaries,keys,values or items

## Iterating over keys 
for key in student.keys():
    print(key)

# Iterating over values 
for values in student.values():
    print(values)

# Iterating over key value pairs 
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
## Nested dictionaries 
students = {
    "student1":{"name":"Divyanshu","age":32},
    "student2":{"name":"Peter","age":35 }
}
print(students)
```

```python
## Accessing nested dictionaries 
students["student1"]["name"]
students["student2"]["age"]
```

```python
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary comprehension 
squares = {x:x*x for x in range(5)}
print(squares)
```

```python
#Conditional dictionary comprehension 
even_squares = {x:x*x for x in range(10) if x%2 ==0}
print(even_squares)
```

```python
## Practical examples of dictionary : 
## Use a dictionary to count frequency of elements in a list 
numbers = [1,2,2,3,3,3,4,4,5,5,6]
frequency = {}
for number in numbers:
    if number in frequency:
        frequency[number] +=1
    else:
        frequency[number] = 1
print(frequency)
```

```python
# Merge two dictionaries into one 
dict1 = {"a":1,"b":2}
dict2 = {"b":3,"c":4}
# dict1 and dict2 are keyword arguments
merge_dict = {**dict1,**dict2}
merge_dict
```
