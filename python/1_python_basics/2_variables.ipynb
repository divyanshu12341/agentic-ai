```python
## Declaring and assigning variables 
age = 32
height = 6.1
name = "Divyanshu"
is_student = True

## Printing variables 
print("age: ",age)
print("height: ",height)
print("Name: ",name)
print("Is student:",is_student)
```

```python
## Naming convention: 
## Variable names should be descriptive 
## Variables should always start with a letter or an underscore(_)
## Variable names are case sensitive 

## Valid variable name: 
first_name = "Divyanshu"
last_name = "Jain"
```

```python
## Invalid variable name:
1name = "Divyanshu" #Starts with a number 
first-name = "Divyanshu" #Contains a hyphen
@name = 'Divyanshu' # Contains a special character 
```

```python
## Understanding variable types 
## Python is dynamically typed means that type of variable determined at runtime. 
age = 25 #integer 
height = 6.1 #float 
name = "Divyanshu" #string
is_student = True #boolean

print(type(age))
print(type(height))
print(type(name))
print(type(is_student))

## Type conversion 
age_str = str(age)
print(type(age_str))
```

```python
age = 25 
print(int(age))
print(type(int(age)))
```

```python
## Dynamic typing: 
## Python allows type of variable to change as program executes.
var = 10
print(var,type(var))

var = "Hello"
print(var,type(var))

var = 3.14
print(var,type(var))
```

```python
## Input 
age = int(input("Enter your age:"))
print("Your age is: ",age,type(age))
```

```python
## Simple calculator :
num1 = float(input("Enter first number:"))
num2 = float(input("Enter second number:"))

sum = num1+num2 
diff = num1-num2
mult = num1*num2
div = num1/num2

print("Sum:",sum)
print("Difference: ",diff)
print("Multiplication: ",mult)
print("Division: ",div)
```
