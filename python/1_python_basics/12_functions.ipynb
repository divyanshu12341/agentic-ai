```python
# Function is block of code that performs specific task 
# It helps in organizing code,reusing code and improving readibility
# syntax: def fun_name(parameters):
# ''docstring ''
# function body 
# return expression 

def oddOrEven(num):
    """ This function finds even or odd """
    if num%2 == 0:
        print("Number is even")
    else:
        print("Number is odd")

## Call this function 
oddOrEven(25)
```

```python
# Function with multiple parameters 

def add(a,b):
    """Add two variables"""
    return a+b

addition = add(2,5)
print(addition)
```

```python
# Default parameters 
def greet(name = "Guest"):
    print(f"Hello {name} welcome to the paradise")

greet("Krissh")
greet()
```

```python
### Variable length Arguments 
### Positional and keyword arguments 

### args --> positional argument
def print_numbers(*args):
    for numbers in args:
        print(numbers)

print_numbers(1,2,3,4,"Krish")


### keywords argument: 
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

print_details(name="Divyanshu",age=26,country="India")

```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f"{val}")
    for key,val in kwargs.items():
        print(f"{key}:{val}")

print_details(1,2,3,4,name="Divyanshu",age=26)
```
