```python
# Exception handling allows us to handle errors gracefully and take actions 
# without stopping program execution 

# Exceptions are events that disrupt normal flow of a program. 
# They occur when error is encountered during program execution 
# Common examples include:
# 1)ZeroDivisionError: Dividing by zero
# 2)FileNotFoundError: File not found
# 3) ValueError: initial value

# try except block 
# Try --> inside this block 
try:
    a=b
except NameError as e:
    print(e)
```

```python
try:
    a = 1
    result = a/0
except ZeroDivisionError as e:
    print(e)
    print("Please enter denominator greater than zero")
```

```python
try: 
    a = 2
    result = a/0 
except Exception as e:
    print(e)
    
```

```python
try:
    num = int(input("Enter a number"))
    result = 10/num
    print(result)
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Enter denominator greater than zero")
except Exception as ex:
    print(ex)
```

```python
#try,except,else block 
try:
    num = int(input("Enter a number"))
    result = 10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Enter denominator greater than zero")
except Exception as ex:
    print(ex)
else:
    print(f"The result is {result}")
```

```python
#try,except,else and finally 
# finally block gets executed whether error is coming or not
try:
    num = int(input("Enter a number"))
    result = 10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("You can't divide by zero")
else:
    print(f"Result is {result}")
finally:
    print("Execution complete")
```

```python
## File handling and exception handling 
try:
    file = open('file.txt','r');
    content = file.read()
    print(content)
except Exception as e:
    print(e)
else:
    print(content)
finally:
    if 'file' in locals() and not file.closed():
        file.close()
        print('File closed')

```
