```python
# Allows to modify behaviour of a function or class method 
# Commonly used to add functionality without modifying original code
# function copy 
def welcome():
    return "Welcome to the advanced python course"
wel = welcome
wel()

del welcome
wel()
```

```python
# Closures functions 

def main_welcome():
    msg = "Welcome"
    def sub_welcome_method():
        print("Welcome to advanced python course")
        print("Please learn these methods properly")
    return sub_welcome_method()
main_welcome()
```

```python
def main_welcomes(func):
    def sub_welcome_method():
        print("Welcome to advance python course")
        func("Welcome everyone in this tutorial")
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
#Decorators
@main_welcomes
def course_introduction():
    print("This is an advanced python course")
```
