```python
# Magic methods in python are predefined methods that you can override 
# to change the behaviour of the objects.
# Start and end with double underscores

'''
__init__
__str__ --> returns string representation of an object
__getitem__ --> Gets an item from the container 
__setitem__ --> Sets an item in the container
 
'''

class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def __str__(self):
        return f"{self.name} {self.age} years old"
    def __repr__(self):
        return f"Person(name = {self.name} age = {self.age})"
    
person = Person("Krish",34)
print(person)
print(repr(person))
```
