```python
# Generators are subclass of iterators 
# Uses yield keyword to produce a series of values lazily which means 
# they can generate values on the fly and do not store them on memory

def square(n):
    for i in range(3):
        yield i**2
square(3)
for i in square(3):
    print(i)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3

gen = my_generator()
next(gen)
```

```python
## Practical example --> reading large files 
# They allow to process one line at a time without loading entire file
#  into memory
def read_large_file(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line

file_path = 'large_file.txt'
for line in read_large_file(file_path):
    print(line.strip())
```
