```python
# Iterators are advanced python concepts which allows efficient looping 
# and memory management. It provides a way to access elements of a 
# collection sequentially without exposing underlying structure.

my_list = [1,2,3,4,5,6]
for i in my_list:
    print(i)
type(my_list)
# we use iter keyword 
# It will not give any element until we iterate iterator
# Internally it uses lazy loading
iterator = iter(my_list)
print(type(iterator))
for i in my_list:
    next(iterator)
```
