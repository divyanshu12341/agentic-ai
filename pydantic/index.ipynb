```python
#Pydantic basically uses for data validation 
#Pydantic basics: Creating and using models 
#They use python data annotations to define structure and 
#validate data at runtime.
from pydantic import BaseModel 
class Person(BaseModel):
    name:str
    age:int
    city:str
person = Person(name = 'Krish',
                age = 35,
                city = 'Bangalore')
type(person)
```

```python
## Model with optional fields 
from typing import Optional 
class Employees(BaseModel):
    id:int
    name:str
    department:str
    salary:Optional[float]=None
    is_active:Optional[bool]=None

emp1 = Employees(id=1,name='John',department='IT')
print(emp1)
emp2 = Employees(id=2,name='John',department='')
```

```python
from pydantic import BaseModel 
from typing import List

class ClassRoom(BaseModel):
    room_number:str
    students:List[str]
    capacity:int
```

```python
#Create a classroom 
classroom = ClassRoom(
    room_number = "A101",
    students = ["Alice","Bob","Charlie"],
    capacity = 30
)
print(classroom)
```

```python
try:
    invalid_val = ClassRoom(room_number = "A1",
                            students = ["Krish",123],
                            capacity=30)
except ValueError as e:
    print(e)
```

```python
from pydantic import BaseModel 
class Address(BaseModel):
    street:str
    city:int
    zip_code:int
class Customer(BaseModel):
    customer_id:int
    name:str
    address:Address

customer = Customer(customer_id = 101,
                    name = "Divyanshu",
                    address = ('111','Jaipur',302021))

```

```python
#Field is used to add constraints
from pydantic import BaseModel,Field
class Item(BaseModel):
    name:str = Field(min_length=2,max_length=50)
    #gt ==> greater than 
    #le ==> less than or equal to 
    price:float = Field(gt=0,le=1000)
    quantity:int = Field(ge=0)

item = Item(name = "Book",price = 1,quantity=10)
print(item)
```
