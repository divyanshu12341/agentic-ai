```python
# Facebook AI Similarity search is a library for efficient similarity search and 
# clustering of dense vectors.
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter
loader = TextLoader("speech.txt")
documents = loader.load()
text_splitter = CharacterTextSplitter(chunk_size=200,chunk_overlap=30)
docs = text_splitter.split_documents(documents)
```

```python
embeddings = OllamaEmbeddings(model = "nomic-embed-text")
db = FAISS.from_documents(docs,embeddings)
db
```

```python
query = "What does the speaker believe about Key Pillars of Viksit Bharat 2047?"
docs = db.similarity_search(query)
docs[0].page_content
```

```python
# As a retriever:
# We can also convert vectorstore db into retriever class
# It allows us to use it in other langchain methods which largely works with retrievers
retriever = db.as_retriever()
retriever.invoke(query)
```

```python
# Similarity search with score 
# Lower score is better 
docs_and_s_score = db.similarity_search_with_score(query)
docs_and_s_score
```

```python
db.save_local("faiss_index")
new_df = FAISS.load_local("faiss_index",embeddings,allow_dangerous_deserialization = True)
new_df
```
